```python
_besd_vars = thinking_and_feeling  + social_processes + practical_issues 
_besd_vars_capital = [f'BeSD{b[4:]}' for b in _besd_vars] 
_besd_description = list(_thinking_and_feeling.values()) + list(_social_processes.values()) + list(_practical_issues.values())
_besd_group = 3*['Thinking feeling'] + 6*['Social processes'] + 7*['Practical issues']
besd_description = pd.DataFrame(columns = ['Variable', 'Description', 'BeSD group'])
besd_description['Variable'] = _besd_vars_capital
besd_description['Description'] = _besd_description
besd_description['BeSD group'] = _besd_group
besd_description = besd_description.set_index('Variable')
```
